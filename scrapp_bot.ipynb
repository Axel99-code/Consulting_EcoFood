```python
base_url = "https://www.guiaoleo.com/buenos-aires/restaurants?page={}"

datos = []
page = 1
max_pages = 5  # Limitar a 10 páginas para evitar sobrecarga
while page <= max_pages:
    response = requests.get(base_url.format(page))
    soup = BeautifulSoup(response.text, "html.parser")

    resto_nombre = soup.find_all("div", class_="flex items-start mb-1") 
    resto_ficha = soup.find_all("div", class_="flex items-center flex-wrap gap-x-2 gap-y-1 text-sm text-gray-600 dark:text-gray-400")
    # si ya no hay más restaurantes, corto el bucle
    if not resto_nombre:
        break

    for i in range(len(resto_nombre)):
        nombre = resto_nombre[i].find("span").get_text(strip=True)
        spans = resto_ficha[i].find_all("span")
        if len(spans) >= 5:
            localizacion = spans[0].get_text(strip=True)
            tipo = spans[2].get_text(strip=True)
            precio = spans[4].get_text(strip=True)
            datos.append([nombre, localizacion, tipo, precio])

    print(f"Página {page} procesada, total acumulado: {len(datos)}")
    page += 1

# convertir a DataFrame
df = pd.DataFrame(datos, columns=["Nombre", "Localización", "Tipo", "Precio"])
print(df)

```
